### Activate Friction Compensation
- Neuromeka provides an option for friction compensation to users of IndySDK. Users can enable this feature according to their preferences via the gRPC Protocol. This feature employs a feedforward friction compensation algorithm that computes joint friction using the desired joint velocity and identified friction parameters.

##### Check friction compensation

```python
control.GetFrictionCompensationState()
```

##### Set friction compensation

```python
enable = True
control.SetFrictionCompensation(enable=enable)
```

##### Turn Off friction compensation

```python
enable = False
control.SetFrictionCompensation(enable=enable)
```

### Activate Customized Controller
- By default, the control mode is set to 0, which corresponds to the default controller.
- To utilize your custom controller you need to switch the control mode.

##### Check the control mode

```python
control.GetCustomControlMode()
```

##### Set control mode

```python
mode = 1
control.SetCustomControlMode(mode=mode)
```

##### Reset to default control mode

```python
mode = 0
control.SetCustomControlMode(mode=mode)
```

### Test Example 
By default, friction compensation is turned off. Initially, the motion is carried out without friction compensation, leading to no friction torque being observed. Subsequently, friction compensation is enabled, and the same motion is performed again. This time, the presence of friction torque is evident in the plot.

```python
mode = 1
control.SetCustomControlMode(mode=mode)
control.GetCustomControlMode()
```
