## JointMove

blending_type = common.Property.BlendingType.OVERRIDE  -> 목표위치 덮어쓰기

blending_type = common.Property.BlendingType.DUPLICATE -> 명령어가 2개가 들어오면 2번째 위치로감....
                                                          같이 들어온 명령어가 동일축에 값이 들어가면 작동하는것처럼 보임

blending_type = common.Property.BlendingType.RADIUS -> 동작 완료후 다음 동작 실행

````python
target_pos1 = [1,1,89,1,89,1]
target_pos2 = [2,2,89,2,89,2]
target_pos3 = [1,1,60,1,60,1]
# target_pos4 = [0,0,0,10,0,0]
# target_pos5 = [0,0,0,0,0,0]
vel_ratio = 50
acc_ratio = 50
for i in range(10):
    time.sleep(0.2)
    control.MoveJ(jpos=target_pos1, blending_type=common.Property.BlendingType.OVERRIDE, vel_ratio=vel_ratio, acc_ratio=acc_ratio)
    time.sleep(0.2)
    control.MoveJ(jpos=target_pos2, blending_type=common.Property.BlendingType.OVERRIDE, vel_ratio=vel_ratio, acc_ratio=acc_ratio)
time.sleep(0.2)
control.MoveJ(jpos=target_pos3, blending_type=common.Property.BlendingType.OVERRIDE, vel_ratio=vel_ratio, acc_ratio=acc_ratio)
print("end")
````
